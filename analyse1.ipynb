```python
import pandas as pd 
# import matplotlib as plt

df = pd.read_csv("paniers.csv")
df.head(2)
```

```python
df.isnull().sum()
```

```python
data = df 
data[["image"]] = data[["image"]].fillna("vide")
data[["image"]]
data.head(1)
```

```python
data[["presentation"]] = data[["presentation"]].fillna("Acunne")
data[["presentation"]].head(2)
```

```python
data[["date_modification"]].to_timestamp
dt = data
db = data
dt[["date_modification"]] = dt[["date_modification"]].fillna(pd.to_datetime("1999-01-01"))
dt["date_modification"] = pd.to_datetime(dt['date_modification']).dt.date
dt[["date_modification"]]
```

```python
db["utilisateur_modification"] = db["utilisateur_modification"].fillna("vide")
db["utilisateur_modification"] 
```
