```python
data[["AUTRE TYPE DE HANDICAP"]] = data[["AUTRE TYPE DE HANDICAP"]].fillna("SANS HANDICAP") 
```

```python
data[["CONTACT 2"]] = data[["CONTACT 2"]].fillna(0)
```

```python
emt = data.groupby("ENTREPRISE").agg(
    total_persones = ("NOM DU BENEFICIAIRE", "count")
)
emt = emt.sort_values("total_persones", ascending= False)
emt.head(17)
```

```python
etb = data.groupby("ETABLISSEMENT FREQUENTE").agg(
    total_persones = ("NOM DU BENEFICIAIRE", "count")
    ).reset_index()
etb = etb.sort_values("total_persones", ascending= False)
etb.head(5)

etb.to_excel("etablissement.xlsx")
```

```python
sexe = data.groupby("SEXE").agg(
    totale_persones = ("NOM DU BENEFICIAIRE", "count")
)
sexe = sexe.sort_values("totale_persones", ascending= False)

sexe.head(4)
sexe.to_excel("sexe.xlsx")
```
